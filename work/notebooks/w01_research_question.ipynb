## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Using the starter dataset, I compare the overall declining rate against the declining rate inside the exact segment my baseline rule targets (stale + still visible pages) — if that segment concentrates more decline than the general population, it's evidence this lane has real signal to work with, before any model is even trained.

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/faja27/flyrank-ml-internship"
REPO_DIR = "flyrank-ml-internship"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)

overall_decline_rate = df["is_declining"].mean()

stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
stale_visible_decline_rate = stale_visible["is_declining"].mean()

print(f"Total pages: {len(df):,} | overall declining rate: {overall_decline_rate:.3f}")
print(f"Stale+visible candidate pages: {len(stale_visible):,} "
      f"({len(stale_visible)/len(df):.1%} of all pages)")
print(f"Declining rate within that segment: {stale_visible_decline_rate:.3f}")
print(f"\nFor reference, the starter pipeline's own verified results show a learned model "
      f"clearly outperforming a hand-written rule on this task "
      f"(Precision@50: baseline 0.240 -> random forest 0.740).")
```
